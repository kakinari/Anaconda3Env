### 正規分布の表示

```python
import numpy as np
import matplotlib.pyplot as plt

# ランダムシードを設定して、再現性のある乱数を生成
np.random.seed(1)

# 平均値と標準偏差を設定して、正規分布に従う乱数を生成
sigma = 2.50
mp = 50
rand_data2 = np.random.normal(mp, sigma, 1_000_000)

# ヒストグラムを描画して、データの分布を可視化
plt.hist(rand_data2, bins=50)
plt.show()

# 平均値と標準偏差を計算して表示
print(rand_data2.mean())
print(rand_data2.std())
```

#### Numpyの演習

```python
import numpy as np

# ランダムシードを設定して、再現性のある乱数を生成
np.random.seed(0)
# 100個の乱数を生成し、10x10の形状に変換
arr1 = np.random.rand(100).reshape(10, 10)
# 生成した配列を表示
print(arr1)
# 各列で0.7より大きい値の数をカウントしてリスト表示
print(np.sum(arr1 > 0.7, axis=0).tolist())
# 各行で0.7より大きい値の数をカウントしてリスト表示
print(np.sum(arr1 > 0.7, axis=1).tolist())
```
